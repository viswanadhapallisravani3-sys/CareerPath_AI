```python
!pip -q install pandas

import pandas as pd
import re

careers = {
    "AI/ML Engineer": {
        "skills": ["python", "machine learning", "statistics", "numpy", "pandas", "deep learning", "nlp", "tensorflow", "pytorch"],
        "interests": ["artificial intelligence", "machine learning", "data science", "nlp"],
        "education": ["computer science", "cse", "aiml", "artificial intelligence", "data science"],
        "activities": ["nptel", "hackathon", "workshop", "project", "internship"],
        "roadmap": ["Python", "NumPy & Pandas", "Statistics", "Machine Learning", "Deep Learning", "NLP", "TensorFlow/PyTorch", "AI Projects"],
        "description": "Builds artificial intelligence and machine learning systems using programming, data and algorithms."
    },
    "Data Scientist": {
        "skills": ["python", "machine learning", "statistics", "pandas", "numpy", "sql", "data visualization", "nlp"],
        "interests": ["data science", "machine learning", "statistics", "artificial intelligence", "data"],
        "education": ["computer science", "cse", "aiml", "data science", "statistics"],
        "activities": ["nptel", "hackathon", "workshop", "project", "internship"],
        "roadmap": ["Python", "Statistics", "SQL", "Pandas", "Data Visualization", "Machine Learning", "Data Science Projects"],
        "description": "Analyzes data and builds machine learning models to solve real-world problems."
    },
    "Data Analyst": {
        "skills": ["python", "sql", "excel", "statistics", "pandas", "data visualization", "power bi", "tableau"],
        "interests": ["data analysis", "data analytics", "business intelligence", "data"],
        "education": ["computer science", "cse", "aiml", "business analytics", "statistics"],
        "activities": ["nptel", "hackathon", "workshop", "project", "internship"],
        "roadmap": ["Excel", "SQL", "Statistics", "Python", "Pandas", "Power BI/Tableau", "Data Analysis Projects"],
        "description": "Uses data analysis and visualization to discover useful insights and support business decisions."
    },
    "Software Developer": {
        "skills": ["python", "java", "c++", "data structures", "algorithms", "git", "problem solving", "oop"],
        "interests": ["software development", "programming", "coding", "software"],
        "education": ["computer science", "cse", "aiml", "information technology"],
        "activities": ["nptel", "hackathon", "workshop", "project", "internship"],
        "roadmap": ["Programming", "OOP", "Data Structures", "Algorithms", "Git/GitHub", "Projects", "Interview Preparation"],
        "description": "Designs, develops and maintains software applications using programming and problem-solving skills."
    },
    "Web Developer": {
        "skills": ["html", "css", "javascript", "react", "node.js", "frontend", "backend", "git"],
        "interests": ["web development", "frontend", "backend", "web design"],
        "education": ["computer science", "cse", "aiml", "information technology"],
        "activities": ["hackathon", "workshop", "project", "internship"],
        "roadmap": ["HTML", "CSS", "JavaScript", "React", "Backend", "Databases", "Web Projects"],
        "description": "Creates websites and web applications using frontend and backend technologies."
    },
    "Cybersecurity Analyst": {
        "skills": ["cybersecurity", "networking", "linux", "ethical hacking", "python", "cryptography", "security"],
        "interests": ["cybersecurity", "ethical hacking", "network security", "security"],
        "education": ["computer science", "cse", "aiml", "cybersecurity", "information technology"],
        "activities": ["nptel", "hackathon", "workshop", "project", "internship"],
        "roadmap": ["Networking", "Linux", "Python", "Cybersecurity Fundamentals", "Ethical Hacking", "Cryptography", "Security Projects"],
        "description": "Protects computer systems, networks and data from security threats."
    },
    "Cloud Engineer": {
        "skills": ["aws", "azure", "cloud computing", "linux", "networking", "python", "docker", "kubernetes"],
        "interests": ["cloud computing", "cloud", "aws", "azure", "infrastructure"],
        "education": ["computer science", "cse", "aiml", "information technology"],
        "activities": ["nptel", "hackathon", "workshop", "project", "internship"],
        "roadmap": ["Linux", "Networking", "AWS/Azure", "Docker", "Kubernetes", "Cloud Security", "Cloud Projects"],
        "description": "Designs and manages cloud infrastructure and cloud-based applications."
    },
    "DevOps Engineer": {
        "skills": ["linux", "git", "docker", "kubernetes", "aws", "python", "ci/cd", "automation"],
        "interests": ["devops", "cloud", "automation", "infrastructure"],
        "education": ["computer science", "cse", "aiml", "information technology"],
        "activities": ["nptel", "hackathon", "workshop", "project", "internship"],
        "roadmap": ["Linux", "Git", "Python", "Docker", "Kubernetes", "CI/CD", "AWS", "DevOps Projects"],
        "description": "Automates software development, testing and deployment processes."
    },
    "UI/UX Designer": {
        "skills": ["figma", "ui design", "ux research", "wireframing", "prototyping", "creativity"],
        "interests": ["ui design", "ux design", "design", "user experience"],
        "education": ["design", "computer science", "cse", "information technology"],
        "activities": ["workshop", "hackathon", "project", "internship"],
        "roadmap": ["Design Fundamentals", "Figma", "UI Design", "UX Research", "Wireframing", "Prototyping", "Portfolio"],
        "description": "Designs user-friendly digital experiences, interfaces and prototypes."
    },
    "Business Analyst": {
        "skills": ["sql", "excel", "statistics", "power bi", "business analysis", "communication"],
        "interests": ["business analysis", "business intelligence", "data", "analytics"],
        "education": ["computer science", "cse", "business analytics", "management"],
        "activities": ["nptel", "hackathon", "workshop", "project", "internship"],
        "roadmap": ["Excel", "SQL", "Statistics", "Power BI", "Business Analysis", "Communication", "Business Projects"],
        "description": "Uses data and business requirements to help organizations make informed decisions."
    }
}

aliases = {
    "py": "python",
    "ml": "machine learning",
    "ai": "artificial intelligence",
    "powerbi": "power bi",
    "reactjs": "react",
    "node": "node.js",
    "cse-aiml": "aiml",
    "cse aiml": "aiml",
    "hackathons": "hackathon",
    "workshops": "workshop"
}

def clean(text):
    text = str(text).lower()
    text = text.replace(",", " ")
    text = re.sub(r"[^a-z0-9+#./ -]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    for old, new in aliases.items():
        text = text.replace(old, new)
    return text

def matches(user_text, items):
    user_text = clean(user_text)
    found = []
    for item in items:
        if clean(item) in user_text:
            found.append(item)
    return found

def calculate(education, skills, interests, activities, experience, data):
    skill_matches = matches(skills, data["skills"])
    interest_matches = matches(interests, data["interests"])
    education_matches = matches(education, data["education"])
    activity_matches = matches(activities, data["activities"])
    experience_matches = matches(experience, data["skills"])

    skill_score = len(skill_matches) / len(data["skills"]) * 100
    interest_score = len(interest_matches) / len(data["interests"]) * 100
    education_score = min(len(education_matches) / len(data["education"]) * 100, 100)
    activity_score = min(len(activity_matches) / len(data["activities"]) * 100, 100)
    experience_score = min(len(experience_matches) / len(data["skills"]) * 100, 100)

    score = (
        skill_score * 0.50 +
        interest_score * 0.20 +
        education_score * 0.15 +
        experience_score * 0.10 +
        activity_score * 0.05
    )

    missing = [x for x in data["skills"] if clean(x) not in [clean(y) for y in skill_matches]]

    return {
        "Score": score,
        "Skill Matches": skill_matches,
        "Interest Matches": interest_matches,
        "Education Matches": education_matches,
        "Activity Matches": activity_matches,
        "Experience Matches": experience_matches,
        "Missing Skills": missing
    }

def get_results(education, skills, interests, activities, experience):
    output = []

    for career, data in careers.items():
        result = calculate(
            education,
            skills,
            interests,
            activities,
            experience,
            data
        )

        output.append({
            "Career": career,
            **result,
            "Roadmap": data["roadmap"],
            "Description": data["description"]
        })

    return sorted(output, key=lambda x: x["Score"], reverse=True)

print("=" * 60)
print("                    CAREERPATH AI")
print("=" * 60)

education = input("Education/Degree: ")
skills = input("Technical skills: ")
interests = input("Interests: ")
activities = input("Certifications/Workshops/Hackathons: ")
experience = input("Projects/Experience: ")

education = clean(education)
skills = clean(skills)
interests = clean(interests)
activities = clean(activities)
experience = clean(experience)

results = get_results(
    education,
    skills,
    interests,
    activities,
    experience
)

print()
print("=" * 60)
print("                 TOP CAREER MATCHES")
print("=" * 60)

for i, result in enumerate(results[:5], 1):
    print(f"{i}. {result['Career']} - {result['Score']:.2f}%")

top = results[0]

print()
print("=" * 60)
print("                     TOP MATCH")
print("=" * 60)

print("Career:", top["Career"])
print(f"Match Score: {top['Score']:.2f}%")

print()
print("Matching Skills:")
print(", ".join(top["Skill Matches"]) if top["Skill Matches"] else "None")

print()
print("Matching Interests:")
print(", ".join(top["Interest Matches"]) if top["Interest Matches"] else "None")

print()
print("Relevant Education:")
print(", ".join(top["Education Matches"]) if top["Education Matches"] else "None")

print()
print("=" * 60)
print("                     SKILL GAP")
print("=" * 60)

if top["Missing Skills"]:
    for skill in top["Missing Skills"]:
        print("-", skill)
else:
    print("No major skill gaps detected.")

print()
print("=" * 60)
print("                PERSONALIZED ROADMAP")
print("=" * 60)

for i, step in enumerate(top["Roadmap"], 1):
    print(f"{i}. {step}")

print()
print("=" * 60)
print("                  CAREER DESCRIPTION")
print("=" * 60)

print(top["Description"])

print()
print("=" * 60)
print("                 CAREER COMPARISON")
print("=" * 60)

table = pd.DataFrame([
    {
        "Career": x["Career"],
        "Score": round(x["Score"], 2),
        "Matching Skills": len(x["Skill Matches"]),
        "Skill Gaps": len(x["Missing Skills"])
    }
    for x in results[:5]
])

print(table.to_string(index=False))

print()
print("=" * 60)
print("             WHAT IF I LEARN A SKILL?")
print("=" * 60)

new_skill = input("Enter a skill you want to learn: ")
new_skill = clean(new_skill)

updated_skills = skills + " " + new_skill

updated_results = get_results(
    education,
    updated_skills,
    interests,
    activities,
    experience
)

old_scores = {x["Career"]: x["Score"] for x in results}

print()
print("Impact of learning:", new_skill)
print()

for x in updated_results[:5]:
    old = old_scores[x["Career"]]
    new = x["Score"]
    change = new - old
    print(f"{x['Career']}: {old:.2f}% -> {new:.2f}% ({change:+.2f})")

print()
print("=" * 60)
print("                  UPDATED TOP MATCH")
print("=" * 60)

print("Career:", updated_results[0]["Career"])
print(f"New Match Score: {updated_results[0]['Score']:.2f}%")

print()
print("=" * 60)
print("             CAREERPATH AI COMPLETE")
print("=" * 60)
```
